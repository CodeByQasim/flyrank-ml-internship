## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
print("ML task type: Clustering")
print("Learning type: Unsupervised learning")
```

1. My lane as an ML task (type)
My lane is a clustering task.

I chose clustering because I do not have a predefined target label that tells me which performance archetype each content page belongs to. Instead, I want the ML method to identify groups of content pages that have similar observable characteristics.

This makes the task an unsupervised learning problem. The goal is not to predict a known label, but to discover and describe meaningful groups in the content inventory.

The final clusters will be used as decision-support rather than treated as true or permanent labels.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
print("Target: None - this is an unsupervised clustering task")
print("Cluster labels will be generated by the clustering method")
```

## 2. Target or proxy

Because this is an unsupervised clustering task, there is **no predefined target label** that the model will predict.

Instead, the model will use selected observable search-performance and content signals as features and group content pages according to their similarity.

The cluster assignment will be a **model-generated grouping**, not an observed outcome or predefined label.

The `trend_direction` field may be useful later for interpreting or checking the resulting clusters, but I will not use it as the target for this clustering task.

Therefore, the main output is the cluster assignment for each content page rather than a predicted target.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
print("Primary clustering metric: Silhouette Score")
print("The score will be calculated after clustering is performed.")
```

## 3. Success metric

I will use the **Silhouette Score** as one quantitative metric for evaluating the clustering.

The Silhouette Score measures how similar each content page is to the pages in its own cluster compared with pages in other clusters.

A higher score generally indicates that the clusters are more separated and internally coherent.

I will not treat a high Silhouette Score alone as proof that the clusters are useful. I will also inspect the size and characteristics of the clusters and consider whether they are interpretable and useful for decision-support.

For this project, the goal is to find clusters that are both quantitatively reasonable and understandable enough to support further investigation.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
print("Number of content items:", len(df))
print("The dataset contains multiple columns that can be investigated as structured signals.")
```

## 5. Why ML beats a fixed rule here

A fixed rule would require me to decide the boundaries between page types in advance, for example by choosing a single threshold for clicks, impressions, or another metric.

The content inventory contains multiple observable signals, and different combinations of these signals may describe different types of pages.

Clustering can consider several features together and identify groups based on similarity rather than relying on one manually chosen threshold.

Therefore, ML may provide a more flexible way to explore the structure of the content inventory.

However, this does not mean that clustering will automatically produce useful groups. The resulting clusters will need to be evaluated for separation, stability, interpretability, and usefulness for decision-support.
